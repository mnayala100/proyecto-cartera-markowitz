# 01. Descarga de datos

Con esto podemos obtener los precios diarios de seis tipos de activo para
construir después la cartera óptima.

Uso ETFs que cotizan todos en **dólares**, para que las rentabilidades
no mezclen el movimiento del tipo de cambio.

| Ticker | Qué es |
|---|---|
| SPY | Acciones de EE. UU. |
| EFA | Acciones de países desarrollados fuera de EE. UU. |
| EEM | Acciones de mercados emergentes |
| AGG | Bonos de EE. UU. |
| GLD | Oro |
| VNQ | Inmobiliario |

**Periodo:** 2016-2025 (10 años de datos diarios).

```python
import yfinance as yf

# Seis tipos de activo con comportamientos distintos
activos = ["SPY", "EFA", "EEM", "AGG", "GLD", "VNQ"]

# Los precios incluyen dividendos y splits,
# así la rentabilidad refleja lo que ganaría un inversor de verdad
precios = yf.download(activos, start="2016-01-01", end="2026-01-01",
                      auto_adjust=True)["Close"]

# Guardo una copia local para no depender de internet en cada ejecución
precios.to_csv("../data/precios.csv")

precios.head()
```
